```python
numbers=[20,55,30,80,45,90,10]
result=[x for x in numbers if x>50]
print(result)
```

```python
numbers=[10,25,35,60,45,15,50,70]
result=[x for x in numbers if x>=20 and x<=50]
print(result)
```

```python
numbers=[10,15,20,30,45,50,60,75]
result=[x for x in numbers if x%3==0 and x%5==0]
print(result)
```

```python
names=["Amit","Rahul","Neha","Rohit","Sneha"]
result=[x for x in names if len(x)>4]
print(result)
```

```python
names=["Amit","Rahul","Ankit","Neha","Arjun","Rohit"]
result=[x for x in names if x.startswith("A")]
print(result)
```

```python
text="programming"
result=[x for x in text if x in "aeiou"]
print(result)
```

```python
data="Python123Programming456"
result=[x for x in data if x.isdigit()]
print(result)
```

```python
numbers=[10,15,20,25,30]
result=["Even" if x%2==0 else "Odd" for x in numbers]
print(result)
```

```python
numbers=[10,-5,20,-8,30,-2]
result=[x if x>0 else 0 for x in numbers]
print(result)
```

```python
prices=[500,1200,800,2000,1500]
result=[x-x*10/100 if x>=1000 else x-x*5/100 for x in prices]
print(result)
```

```python
employees=[{"name":"Amit","salary":45000},{"name":"Rahul","salary":60000},{"name":"Neha","salary":75000},{"name":"Rohit","salary":35000}]
result=[x["name"] for x in employees if x["salary"]>50000]
print(result)
```

```python
products=[499,1299,799,2499,999,1599]
result=[x for x in products if x>1000]
print(result)
```

```python
prices=[500,1200,2500,3000,1500,5000]
result=[x-x*20/100 for x in prices if x>2000]
print(result)
```
